## Part B — The vendor's sampling proposal (5 min) 
The vendor wants to record **one sample every 25 hours**. Traffic has a strong **24-hour cycle**.

**B1.** Implement:
- `nyquist_max_interval(period_h)`: the **largest** sampling interval that avoids aliasing (Nyquist: $f_s > 2f$)
- `aliased_period(true_period, sample_interval)`: the **apparent period**, using aliased frequency $f_a = |f - f_s|$ for $f_s/2 < f < 3f_s/2$ (the case in this activity)

Then compute `apparent_period_25h` for the vendor's plan.

```python
def nyquist_max_interval(period_h):
    return period_h / 2          # f_s > 2f  ->  interval < period / 2

def aliased_period(true_period, sample_interval):
    f, fs = 1 / true_period, 1 / sample_interval
    fa = abs(f - fs)
    return 1 / fa

apparent_period_25h = aliased_period(24, 25)

check_B(nyquist_max_interval, aliased_period, apparent_period_25h)
```

```python
# (given) What the vendor's data would look like for spring 2017
seg = tr["traffic_volume"].loc["2017-03-01":"2017-05-31"].asfreq("h").interpolate(limit=3)
fig, ax = plt.subplots(figsize=(14, 3.5))
ax.plot(seg.index, seg.values, c="lightgrey", lw=0.4, label="true hourly")
ax.plot(seg.index[::25], seg.values[::25], "-o", c="red", ms=3, label="vendor: every 25 h")
ax.set_title(f"Sampling every 25 h turns the 24-h cycle into a fake ~{apparent_period_25h/24:.0f}-day wave"); ax.legend(); plt.show()
```

**B2. A manager looks only at the vendor's 25-hour samples and says: *'Traffic rises and falls on a roughly 25-day cycle — let's schedule closures in the low phase.'* Explain what is really happening, using the words sampling rate, Nyquist and aliasing.**

**Answer:** The manager is seeing an artefact. The true cycle is 24 h, but sampling every 25 h is far below the required sampling rate: the **Nyquist** rule needs an interval shorter than 12 h (more than 2 samples per cycle). Undersampling makes the 24-h cycle **alias** into a slow fake wave with a 600 h (25-day) period, so each sample lands one hour later in the daily cycle and the sequence drifts through the whole cycle; the '25-day cycle' does not exist in the traffic.

**B3. What maximum sampling interval would you recommend to MnDOT? Would sampling exactly every 12 hours be safe? Justify briefly.**
**Answer:** I would choose an interval comfortably shorter than 12 hours; keeping hourly samples is a practical option, because the rush-hour peaks are narrow and have a higher-frequency content than a plain 24-h sine wave. Sampling exactly every 12 h is not safe: Nyquist requires *strictly* more than two samples per cycle, and at exactly 2 per cycle the result depends on phase (the samples could miss the peaks or hit the same part of the cycle every time), and it cannot resolve the morning and evening peaks.

**C2. In the zoom, the trailing average dips *after* 4 July while the centred one dips *around* it. How many days does a 7-day trailing average lag, and why does that matter to someone reading the chart?**

**Answer:** A 7-day trailing average uses the last 7 days, so it is centred 3 days earlier and lags by **3 days**. It matters because the dip appears after the event, so a reader may think the holiday effect happened days later, or mis-time a cause and effect. A centred average aligns with the event but cannot be computed for the most recent days.

**C3. The engineers care about which *day of the week* is quietest. Which is the right view: raw daily, 7-day MA, or 30-day MA? What does each larger window hide? Also interpret your `points_per_pixel` value.**
**Answer:** Raw daily values (or daily values grouped by weekday) are the right view, because a 7-day moving average averages across exactly one week and so removes the day-of-week pattern, and a 30-day average also hides holidays and short events, showing only slow seasonal change. The `points_per_pixel` value (about 8–9) means each pixel column would have to draw several hourly points on top of each other, causing overplotting, so the data must be aggregated or smoothed before plotting.

## Part E — Design the animated *Traffic Pulse* for the lobby screen (6 min) 
**Design brief:** the screen animates **24 hourly states** of a typical weekday. Each hour is shown for **0.75 s** at **30 fps**, with a smooth transition between hours.

💻 **E1.** Implement and compute:
- `frame_budget_ms(fps)`: milliseconds available per frame
- `smoothstep(u)`: the ease-in-out function $3u^2-2u^3$
- `n_frames`, `duration_s`: for the brief above
- `compression`: seconds of **data** per second of **animation** (one hour of data per 0.75 s)

```python
def frame_budget_ms(fps):
    return 1000 / fps

def smoothstep(u):
    return 3 * u**2 - 2 * u**3

duration_s  = 24 * 0.75            # 18 s
n_frames    = int(duration_s * 30) # 540 frames at 30 fps
compression = 3600 / 0.75          # 4800 s of data per second of animation

print(f"frame budget @30fps: {frame_budget_ms(30):.1f} ms | duration {duration_s} s | frames {n_frames} | compression {compression:.0f}:1")
print(f"smoothstep(0.25) = {smoothstep(0.25):.5f}  (linear would be 0.25)")
```

💻 **E2.** A second version shows how the **weekday profile changes month by month** in 2017. Build `fig` with `px.bar` using **`animation_frame="Month"`**, and **fix the y-axis** so bars stay comparable across frames (congruence).

```python
# (given) monthly weekday profiles for 2017
wk17 = wk.loc["2017"]
mp = wk17.groupby([wk17.index.month.rename("month"), wk17.index.hour.rename("hour")])["traffic_volume"].mean().rename("vehicles").reset_index()
mp["Month"] = mp["month"].map(lambda m: calendar.month_abbr[m])
MONTHS = [calendar.month_abbr[i] for i in range(1, 13)]
mp.head()
```

```python
fig = px.bar(mp, x="hour", y="vehicles",
             animation_frame="Month", category_orders={"Month": MONTHS}, range_y=[0, 7500],
             title="Traffic Pulse 2017 — weekday profile by month")
fig.show()

check_E(frame_budget_ms, smoothstep, n_frames, duration_s, compression, fig)
```

```python
# (given) The SAME data as small multiples — compare with the animation
fig2 = px.bar(mp, x="hour", y="vehicles", facet_col="Month", facet_col_wrap=6, category_orders={"Month": MONTHS}, height=420,
              title="Small multiples: all 12 months at once")
fig2.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1])); fig2.show()
```

**E3. Play the animation and try to answer: *In which month is the 17:00 peak lowest?* Then answer it with the small multiples. Which was faster and more certain, and why? Use the terms working memory and 'eyes beat memory'.**
**Answer:** Using the animation took longer and gave a less certain answer: to find the month with the lowest 17:00 bar I had to remember each frame's 17:00 height while it changed, which overloads **working memory** and I would probably miss the minimum between frames. With the small multiples all 12 months are visible together and I can compare the 17:00 bars directly: **eyes beat memory**, so it was faster and more certain.

**E4. What would go wrong if you left out `range_y` (letting each frame auto-scale)? Which Heer & Robertson principle does that violate?**
**Answer:** Without `range_y`, each frame would rescale its own y-axis, so a low month and a high month would look the same height and the axis labels would change from frame to frame; viewers could not compare months. This violates the **congruence** principle of Heer & Robertson (the animated representation should preserve the meaning of the data it shows, here a consistent scale).

**E5. Why use smoothstep (slow-in/slow-out) between hours instead of linear motion or a jump cut? Refer to your `smoothstep(0.25)` value and the per-frame jumps at the start and end.**

**Answer:** `smoothstep(0.25) = 0.156` is below the linear value of 0.25, so the motion starts slowly (it covers only 16% of the change in the first quarter of the time), speeds up in the middle and slows again at the end. With about 22 frames per hour-to-hour transition, the first-frame jump is about 0.6% of the change (against about 4.5% for linear motion) and the middle jumps are about 1.5x the linear ones. Slow-in/slow-out lets the eye track the bars and see where they came from and go, whereas linear motion has abrupt starts and stops that feel mechanical, and a jump cut gives no continuity so the viewer cannot tell which bar became which.

Write your recommendation to your manager.

**F1. For the **road-works engineers**: which visualization(s) should they get, and should they be animated? Name at least two idioms from the lecture and justify.**

**F2. For the **public lobby screen**: animation is acceptable here. Give three concrete design rules you'd apply (timing, fps, scales, labels, playback) so the public actually perceives the change.**

**Answer:** Rules for the lobby screen: (1) **timing**: hold each hour long enough to read, about 0.75 s, and use slow-in/slow-out easing, with a pause at the evening peak; (2) **fps and smoothness**: at least 30 fps (a budget of about 33 ms per frame) with a smooth transition between states; (3) **scales and labels**: keep the y-axis fixed across frames (congruence), keep a visible clock/month label and a stable colour per element, and mark the peak so the public knows what to look at; (4) **playback**: loop, with the option to pause or replay, and show small multiples or a trace of the previous frame so changes can be seen.

**F3. A viewer watching the lobby animation fails to notice that the evening peak disappears in one month. What is this phenomenon, and name two ways to reduce it.**

**Answer:** This is **change blindness**: with many things moving, a change in one bar is missed if attention is elsewhere. Reduce it by (a) highlighting or annotating the changing element (e.g. flash or outline the evening peak and use a ghost of the previous frame), (b) slowing the animation or pausing at key frames, (c) showing small multiples or a side-by-side comparison instead of relying on memory, or (d) animating fewer objects at once.
